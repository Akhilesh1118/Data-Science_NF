```python
import pandas as pd

Df1 = pd.DataFrame({
    "A": [1, 2, 3, 4],
    "B": ["A", "B", "C", "D"]
})

Df2 = pd.DataFrame({
    "A": [1, 2, 3, 4],
    "D": ["a", "b", "c", "d"]
})
```

```python
pd.merge(Df1, Df2)
```

```python
import pandas as pd

Df1 = pd.DataFrame({
    "A": [1, 2, 3, 4],
    "B": ["A", "B", "C", "D"]
})

Df2 = pd.DataFrame({
    "A": [1, 2, 3, 4],
    "B": ["a", "b", "c", "d"]
})
```

```python
pd.merge(Df1, Df2, on = "A" , how="inner")
```

```python

pd.merge(Df1, Df2, on = "A" , how="outer")
```

```python
pd.merge(Df1, Df2, on = "A" , how="left")
```

```python
pd.merge(Df1, Df2, on = "A" , how="right")
```

```python
Df1 = pd.DataFrame({
    "A": [1, 2, 3, 4],
    "B": ["A", "B", "C", "D"]
})

Df2 = pd.DataFrame({
    "A": [1, 5, 7, 3],
    "B": ["a", "b", "c", "d"]
})
```

```python
pd.merge(Df1, Df2, on = "A" , how="inner")
```

```python
pd.merge(Df1, Df2, on = "A" , how="outer")
```

```python
pd.merge(Df1, Df2, on = "A" , how="left")
```

```python
pd.merge(Df1, Df2, on = "A" , how="right")
```

```python
Df1 = pd.DataFrame({
    "A": [1, 2, 3, 4,5],
    "B": ["A", "B", "C", "D", "E"]
})

Df2 = pd.DataFrame({
    "A": [1, 2, 3, 4,5],
    "B": ["a", "b", "c", "d","e"]
})
```

```python
pd.merge(Df1, Df2, left_index = True, right_index= True)
```

```python
pd.merge(Df1, Df2, left_index = True, right_index= True , suffixes=("_2019", "_2020"))
```

```python
pd.merge(Df1, Df2, on = "A" , how="inner", indicator=True)
```
