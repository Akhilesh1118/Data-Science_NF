Joining DataFrames
What is join() in Pandas ?
- join() is used to combiine two DataFrames based on their index
- It is simpler than merge () when data is already indexed
- Mainly used when :
> One Data frame is primary \
> Aother DataFrame adds extra columns


```python
import pandas as pd

Df1 = pd.DataFrame({"A": [1, 2, 3, 4], "B": ["A", "B", "C", "D"]})

Df2 = pd.DataFrame({ "C": [1, 2, 3, 4], "D": ["a", "b", "c", "d"]})
```

```python
Df1 = pd.DataFrame({"A": [1, 2, 3, 4], "B": ["A", "B", "C", "D"]})

Df2 = pd.DataFrame({ "C": [1, 2, 3, 4], "D": ["a", "b", "c", "d"]})
```

Index in join .


```python
Df1 = pd.DataFrame({"A": [1, 2, 3, 4], "B": ["A", "B", "C", "D"]}, index=["a" , "b" , 2, "d"])

Df2 = pd.DataFrame({ "C": [1, 2, 3, 4], "D": ["a", "b", "c", "d"]})
Df1.join(Df2)
```

```python
Df1 = pd.DataFrame({"A": [1, 2, 3, 4], "B": ["A", "B", "C", "D"]}, index=["a" , "b" , 2, "d"])

Df2 = pd.DataFrame({ "C": [1, 2, 3, 4], "D": ["a", "b", "c", "d"]}, index=["a",2,"d" , "b"])
Df1.join(Df2)
```

```python
Df1 = pd.DataFrame({"A": [1, 2, 3, 4], "B": ["A", "B", "C", "D"]}, index=["a" , "b" , 2, "d"])

Df2 = pd.DataFrame({ "C": [1, 2], "D": ["a", "b"]}, index=["c" , "d" ])

Df1.join(Df2,how="inner")
```

```python
Df1 = pd.DataFrame({"A": [1, 2, 3, 4], "B": ["A", "B", "C", "D"]}, index=["a" , "b" , 2, "d"])

Df2 = pd.DataFrame({ "C": [1, 2], "D": ["a", "b"]}, index=["c" , "d" ])

Df1.join(Df2,how="outer")
```

```python
Df1 = pd.DataFrame({"A": [1, 2, 3, 4], "B": ["A", "B", "C", "D"]}, index=["a" , "b" , 2, "d"])

Df2 = pd.DataFrame({ "C": [1, 2], "D": ["a", "b"]}, index=["c" , "d" ])

Df1.join(Df2,how="left")
```

```python
Df1 = pd.DataFrame({"A": [1, 2, 3, 4], "B": ["A", "B", "C", "D"]}, index=["a" , "b" , 2, "d"])

Df2 = pd.DataFrame({ "C": [1, 2], "D": ["a", "b"]}, index=["c" , "d" ])

Df1.join(Df2,how="right" )
```
