```python
# PANDAS SERIES
import pandas as pd
l1=[2,4,6,8,10]
var=pd.Series(l1)
print(var)
print(type(var))
```

```python
l1=[2,4,6,8,10]
var=pd.Series(l1,index=["a","bYE","3","d","akhilesh"])     # CUSTOMIZED INDEX
print(var)
```

```python
l1=[2,4,6,8,10]
var=pd.Series(l1,index=["a","bYE","3","d","akhilesh"],dtype="float")     # CUSTOMIZED value type from integer to float
print(var)
```

```python

l1=[2,4,6,8,10]
var=pd.Series(l1,index=["a","bYE","3","d","akhilesh"],dtype="float", name="akhilesh") # adding name
print(var)
```

```python
var2=pd.Series(5, index=[1,2,3,4,5])     # it's like (5) * [1,2,3,4,5]= [5,5,5,5,5] at [1,2,3,4,5]
print(var2)
```

```python
# series operations
myseries=[2,3,4,5,6]
v=pd.Series(myseries)
sum=v.sum()
max=v.max()
min=v.min()
print(sum)
print(max)
print(min)
print(v)
```
