```python
import pandas as pd
```

```python
# Merging & Joining
info = pd.DataFrame({
"name": ["Saurav", "Kuldeep", "Akash", "Gaurav"],
"roll_no": [101, 102, 103, 104],
})
marks = pd.DataFrame({
"marks": [92, 90, 85, 96],
"roll_no": [101, 102, 103, 104],
"Dept": ["CSE", "CSE", "CDS", "CSE"]

})
```

```python

pd.merge(info, marks, on="roll_no")    # Inner Join
```

```python
pd.merge(info, marks, on="roll_no", how="left")   # Left Join
```

```python
pd.merge(info, marks, on="roll_no", how="right")  # Right Join
```

```python
pd.merge(info, marks, on="roll_no", how="outer") # Outer Join
```
