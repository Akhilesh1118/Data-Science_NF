# Line Plot

```python
import matplotlib.pyplot as plt
from matplotlib.lines import lineStyles
```

```python
# Integrate Data set
age = [18,20,23,21,25]
salary = [15000,18000,20000,19000,30000]

# plot graph
plt.plot(age,salary)
plt.show()
```

```python
age = [18,20,23,21,25]
salary = [15000,18000,20000,19000,30000]

# add labels and title in charts
plt.plot(age,salary)
plt.xlabel("Age of Employee")
plt.ylabel("Salary of Employee")
plt.title("Age vs Salary")

plt.show()
```

```python
age = [18,20,23,21,25]
salary = [15000,18000,20000,19000,30000]

#adjust figure
plt.figure(figsize=(10,5))

# plot line chart
plt.plot(age,salary, marker='o',linestyle = "--" ,linewidth = "2") # linestyle = --, _, .-: , -.

# add labels adn title in charts
plt.xlabel("Age of Employee" , fontsize = 12 , color= "blue")
plt.ylabel("Salary of Employee", fontsize = 12 ,color="green")
plt.title("AGE VS SALARY " , fontsize = 18 , color = "BLACK" )

# add grids
plt.grid(True)

# Show the charts .
plt.show()
```

```python
# Data
age = [18, 20, 23, 21, 25]
salary = [15000, 18000, 20000, 19000, 30000]

# Create figure
plt.figure(figsize=(12, 6))

# Plot line chart
plt.plot(
    age,
    salary,
    marker='o',
    markersize=10,
    linestyle='--',
    linewidth=3,
    color='darkblue',
    markerfacecolor='orange',
    markeredgecolor='black',
    label='Salary Trend'
)

# Labels and title
plt.xlabel("Age of Employee", fontsize=14, color="blue", fontweight="bold")
plt.ylabel("Salary of Employee (₹)", fontsize=14, color="green", fontweight="bold")
plt.title("Age vs Salary Analysis", fontsize=20, color="darkred", fontweight="bold")

# Grid
plt.grid(True, linestyle=':', alpha=0.7)

# Highlight highest salary point
max_salary = max(salary)
max_index = salary.index(max_salary)

plt.scatter(
    age[max_index],
    max_salary,
    color='red',
    s=200,
    zorder=5,
    label='Highest Salary'
)

# Annotate all points
for x, y in zip(age, salary):
    plt.annotate(
        f"₹{y}",
        (x, y),
        textcoords="offset points",
        xytext=(0, 10),
        ha='center',
        fontsize=10
    )

# Customize ticks
plt.xticks(age, fontsize=11)
plt.yticks(fontsize=11)

# Background color
plt.gca().set_facecolor('#f5f5f5')

# Legend
plt.legend()

# Tight layout
plt.tight_layout()

# Show chart
plt.show()
```

# Histogram

```python
# Integrate Data set
marks = [50,56,72,88,98,60,62,74,92,90]

# plot graph
plt.hist(marks)
plt.show()
```

```python
from matplotlib.lines import lineStyles
# Integrate Data set
marks = [50,56,72,88,98,60,62,74,92,90]

# plot graph
plt.hist(marks)
plt.hist(marks , bins = 5 , edgecolor='brown' , color='beige' , alpha = 0.9)

# give title
plt.title("marks distribution")

# add labels adn title in charts
plt.xlabel("Marks")
plt.ylabel("Frequency")

# add grids
plt.grid(axis="y" , color='tan' , linestyle='--' )

plt.show()
```

```python
from matplotlib.lines import lineStyles
# Integrate Data set
marks = [10,20,30,40,50,60,70,80]

#adjust figure
plt.figure(figsize=(8,5))

# plot graph
plt.hist(marks)
plt.hist(marks , bins = 5 , edgecolor='brown' , color='beige' , alpha = 0.9)

# give title
plt.title("marks distribution")

# add labels adn title in charts
plt.xlabel("Marks")
plt.ylabel("Frequency")

# add grids
plt.grid(axis="y" , color='tan' , linestyle='--' )

plt.show()
```

# Bar Chart - Category comparision

2. Grouped Bar Chart

```python
# Integrate Data set
subject = ["Python" , "C", "java" , "C++"]
marks = [50,56,72,88]


#adjust figure
plt.figure(figsize=(8,5))

# plot graph
plt.bar(subject , marks , color='tan'   )

# give title
plt.title("Score")

# add labels adn title in charts
plt.xlabel("subject")
plt.ylabel("Marks")

# add the value in the top of bar
for index, value in enumerate(marks):
    plt.text(index, value+1, str(value), ha='center')
plt.show()
```
