## Chapter 1 Data Visualization with Matplotlib

What is Data Visualization & Why  it Matters

Data Visualization Using Matplotlib

# 1. What is a line plot?
A line chart or line plot is a graphical representation used to show the

 relationship between two continuous variables by connecting data points with

 a straight line. It is commonly used to


* used to show trend over time or order
* connects points with a line
* best for continuous data
   



# 2. Basic  line plot

```python
# import libraries
import matplotlib.pyplot as plt
import numpy as np
```

```python
age = [18, 20, 23, 21, 25]
salary = [15000, 18000, 20000, 19000, 30000]

plt.plot(age, salary)
plt.show()
```

## 3) Add title and axis labels

```python
plt.plot(age, salary)
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary")
plt.show()
```

# 4. Add style options

```python
plt.plot(age, salary, marker='o', linestyle='--', linewidth=2, color='blue')
plt.show()
```

* marker = point style
* linestyle = line pattern
* linewidth = line thickness
* grid() = helps read values





# 5. Add grid and figure size

```python
plt.figure(figsize=(10, 5))
plt.plot(age, salary, marker='o', linestyle='--', linewidth=2)
plt.grid(True)
plt.show()
```

# 6. Add legend and multiple lines

```python
salary1 = [15000, 18000, 20000, 19000, 30000]
salary2 = [12000, 16000, 21000, 22000, 28000]

plt.plot(age, salary1, marker='o', label="Company A")
plt.plot(age, salary2, marker='s', label="Company B")
plt.legend()
plt.show()
```

# 7. Highlight a special point
Use scatter() to mark max/min point.

```python

plt.figure(figsize=(10, 6))
plt.plot(age, salary, marker='o', linestyle='--', color='blue', label='Salary Trend')

# Highlight max point
max_salary = max(salary)
max_age_index = salary.index(max_salary)
plt.scatter(age[max_age_index], max_salary, color='red', s=100, zorder=5, label=f'Max Salary: {max_salary}')

# Highlight min point
min_salary = min(salary)
min_age_index = salary.index(min_salary)
plt.scatter(age[min_age_index], min_salary, color='green', s=100, zorder=5, label=f'Min Salary: {min_salary}')

plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary with Max/Min Highlight")
plt.grid(True)
plt.legend()
plt.show()
```

# 8. Annotate values on the chart

```python
plt.plot(age, salary, marker='o', linestyle='--', color='blue', label='Salary Trend')

for x, y in zip(age, salary):
    plt.annotate(f"{y}", (x, y), textcoords="offset points", xytext=(0, 10), ha='center')

plt.legend()
plt.show()
```

# 9. Fill area under the line

```python
plt.plot(age, salary, marker='o')
plt.fill_between(age, salary, alpha=0.2)
plt.show()
```

# 10. Use subplots for comparison

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].plot(age, salary)
ax[0].set_title("Salary Trend 1")

ax[1].plot(age, [x * 1.1 for x in salary])
ax[1].set_title("Salary Trend 2")

plt.tight_layout()
plt.show()
```

# 11. Use a second y-axis
This is advanced and useful when two values have different scales.

```python
fig, ax1 = plt.subplots()

ax1.plot(age, salary, color='blue', label='Salary')
ax1.set_ylabel("Salary", color='blue')

ax2 = ax1.twinx()
ax2.plot(age, [10, 12, 15, 14, 18], color='red', label='Other Value')
ax2.set_ylabel("Other Value", color='red')

plt.show()
```

```python
# Data
age = [18, 20, 23, 21, 25]
salary = [15000, 18000, 20000, 19000, 30000]

# Create figure
plt.figure(figsize=(12, 6))

# Plot line chart
plt.plot(
    age,
    salary,
    marker='o',
    markersize=10,
    linestyle='--',
    linewidth=3,
    color='darkblue',
    markerfacecolor='orange',
    markeredgecolor='black',
    label='Salary Trend'
)

# Labels and title
plt.xlabel("Age of Employee", fontsize=14, color="blue", fontweight="bold")
plt.ylabel("Salary of Employee (₹)", fontsize=14, color="green", fontweight="bold")
plt.title("Age vs Salary Analysis", fontsize=20, color="darkred", fontweight="bold")

# Grid
plt.grid(True, linestyle=':', alpha=0.7)

# Highlight highest salary point
max_salary = max(salary)
max_index = salary.index(max_salary)

plt.scatter(
    age[max_index],
    max_salary,
    color='red',
    s=200,
    zorder=5,
    label='Highest Salary'
)

# Annotate all points
for x, y in zip(age, salary):
    plt.annotate(
        f"₹{y}",
        (x, y),
        textcoords="offset points",
        xytext=(0, 10),
        ha='center',
        fontsize=10
    )

# Customize ticks
plt.xticks(age, fontsize=11)
plt.yticks(fontsize=11)

# Background color
plt.gca().set_facecolor('#f5f5f5')

# Legend
plt.legend()

# Tight layout
plt.tight_layout()

# Show chart
plt.show()
```
