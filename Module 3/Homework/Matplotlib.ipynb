# 1. Pie Chart

```python
# Pie chart of revenue by month
plt.pie(df["Revenue"], labels=df["Months"], autopct="%1.1f%%")

plt.title("Revenue Distribution by Month")
plt.show()
```

# 2. Bar Graph

```python
# Bar graph showing units sold each month
plt.bar(df["Months"], df["Unit"])

plt.title("Units Sold by Month")
plt.xlabel("Months")
plt.ylabel("Units Sold")

plt.show()
```

# 3. Histogram

```python
# Histogram of revenue values
plt.hist(df["Revenue"], bins=5)

plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")

plt.show()
```

#4. Line Chart

```python
# Line chart showing revenue trend
plt.plot(df["Months"], df["Revenue"])

plt.title("Monthly Revenue Trend")
plt.xlabel("Months")
plt.ylabel("Revenue")

plt.show()
```

# 5. Scatter Plot

```python
import matplotlib.pyplot as plt

# Relationship between units sold and revenue
plt.scatter(df["Unit"], df["Revenue"])

plt.title("Units Sold vs Revenue")
plt.xlabel("Units Sold")
plt.ylabel("Revenue")

plt.show()
```
